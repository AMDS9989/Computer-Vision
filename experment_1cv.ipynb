```python
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open("tree.png")

# Convert image to grayscale
gray_img = img.convert("L")

# Display original image
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

# Display grayscale image
plt.subplot(1, 2, 2)
plt.imshow(gray_img, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("3.png.jpg")

# Convert image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Canny edge detection
edges = cv2.Canny(gray, 100, 200)

# Display original image
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# Display Canny outline
plt.subplot(1, 2, 2)
plt.imshow(edges, cmap="gray")
plt.title("Canny Outline")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("4.jpg")

# Create a kernel
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

# Dilate the image
dilated = cv2.dilate(img, kernel, iterations=1)

# Display original image
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# Display dilated image
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(dilated, cv2.COLOR_BGR2RGB))
plt.title("Dilated Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("5.jpg")

# Create a kernel
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

# Erode the image
eroded = cv2.erode(img, kernel, iterations=1)

# Display original image
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# Display eroded image
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(eroded, cv2.COLOR_BGR2RGB))
plt.title("Eroded Image")
plt.axis("off")

plt.show()
```

```python
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt

# Read the image
img = Image.open("2.png.jpg")

# Apply Gaussian Blur
blur_img = img.filter(ImageFilter.GaussianBlur(5))

# Display original image
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

# Display blurred image
plt.subplot(1, 2, 2)
plt.imshow(blur_img)
plt.title("Gaussian Blurred Image")
plt.axis("off")

plt.show()
```
